### Explicación de las Clases, Herencia y Métodos

### Explicación de Clases y Herencia
El proyecto implementa tres clases aplicando **Herencia** y **Polimorfismo**:

* **Clase Padre: `Evaluacion`** (Calificación genérica)
  * `__init__`: Guarda el nombre y la nota base.
  * `obtener_informacion()`: Retorna un texto limpio con estos datos.

* **Clase Hija: `Examen`** (Hereda de `Evaluacion`)
  * `__init__`: Usa `super()` para heredar los datos base y añade el atributo `tipo_examen` (ej. Parcial).
  * `obtener_informacion()`: Sobrescribe el método del padre para incluir el tipo de examen.

* **Clase Hija: `TrabajoPractico`** (Hereda de `Evaluacion`)
  * `__init__`: Usa `super()` e incorpora el atributo booleano `es_grupal`.
  * `obtener_informacion()`: Sobrescribe el método para especificar si la tarea fue grupal o individual.  

```python

# 1. CLASE PADRE
class Evaluacion:
    def __init__(self, nombre, nota):
        self.nombre = nombre
        self.nota = nota

    def obtener_informacion(self):
        return f"{self.nombre}: {self.nota:.2f} pts"


# 2. CLASE HIJA (Herencia)
class Examen(Evaluacion):
    def __init__(self, nombre, nota, tipo_examen):
        super().__init__(nombre, nota)  # Llama al constructor del padre
        self.tipo_examen = tipo_examen

    def obtener_informacion(self):
        # Sobrescribe el método del padre agregando su propio formato
        return f"{self.nombre} (Examen {self.tipo_examen}): {self.nota:.2f} pts"


# 3. CLASE HIJA (Herencia)
class TrabajoPractico(Evaluacion):
    def __init__(self, nombre, nota, es_grupal):
        super().__init__(nombre, nota)
        self.es_grupal = es_grupal

    def obtener_informacion(self):
        modalidad = "Grupal" if self.es_grupal else "Individual"
        return f"{self.nombre} (Trabajo {modalidad}): {self.nota:.2f} pts"

print("Clases definidas correctamente en memoria")
```

### Instanciación y Datos Precargados
Inicializamos el sistema con una lista que contiene objetos ya creados a partir de las clases hijas.

```python
# Carga inicial de datos usando Objetos
registro_academico = [
    TrabajoPractico("Tarea 1", 85.0, es_grupal=False),
    Examen("Examen Parcial 1", 68.0, tipo_examen="Parcial"),
    TrabajoPractico("Sistemático 1", 92.0, es_grupal=True)
]

print(" Datos precargados inicializados exitosamente.")
```

### Función 1: Mostrar Notas Registradas
Recorre la lista de objetos `registro_academico` e imprime la información de cada evaluación. Al llamar a `obtener_informacion()`, el programa decide automáticamente (polimorfismo) el formato de texto correcto dependiendo de si es un Examen o un Trabajo Práctico.

```python
def mostrar_notas_precargadas(registro):
    print("\n--- NOTAS REGISTRADAS EN LA ASIGNATURA ---")
    if not registro:
        print("No hay evaluaciones registradas.")
        return

    for i, evaluacion in enumerate(registro):
        print(f"{i + 1}. {evaluacion.obtener_informacion()}")
```


### Fnción 2: Registrar Nueva Evaluación
Captura los datos del usuario para crear un nuevo registro. Implementa un bloque try-except para validar que la calificación sea estrictamente numérica y se encuentre en el rango de 0 a 100. Posteriormente, instancia un objeto Examen o TrabajoPractico según lo que el usuario haya seleccionado.

```python
def registrar_nota(registro):
    print("REGISTRAR NUEVA EVALUACIÓN")
    print("Tipo de actividad:")
    print("1. Examen")
    print("2. Trabajo Práctico")
    
    try:
        opcion_tipo = int(input("Seleccione el tipo (1 o 2): "))
    except ValueError:
        print(" Error: Selección inválida. Se registrará como Trabajo Práctico por defecto.")
        opcion_tipo = 2

    nombre_evaluacion = input("Ingrese el nombre de la evaluación: ").strip()

    if not nombre_evaluacion:
        print("Error: El nombre no puede estar vacío.")
        return

    while True:
        try:
            nota_input = float(input("Ingrese la nota obtenida (0 - 100): "))

            if 0 <= nota_input <= 100:
                if opcion_tipo == 1:
                    nueva_evaluacion = Examen(nombre_evaluacion, nota_input, "General")
                else:
                    nueva_evaluacion = TrabajoPractico(nombre_evaluacion, nota_input, es_grupal=False)
                
                registro.append(nueva_evaluacion)
                print(f"✓ Éxito: Evaluación guardada correctamente con {nota_input:.2f} pts.")
                break
            else:
                print(" Error: La nota debe estar entre 0 y 100. Intente de nuevo.")

        except ValueError:
            print(" Error: Por favor, ingrese un número válido (ej. 85 o 78.5).")
```

### Función 3: Calcular Promedio y Estado
Extrae las calificaciones de todos los objetos guardados en la lista utilizando comprensión de listas, calcula el promedio aritmético simple y evalúa mediante estructuras condicionales (if-elif-else) si el estado final es Aprobado, En Riesgo o Reprobado.


```python
def calcular_promedio_y_estado(registro):
    print("BALANCE ACADÉMICO Y ESTADO ")
    if not registro:
        print("No hay datos para calcular el promedio.")
        return

    # Extrae las notas de los objetos para operarlas
    lista_notas = [evaluacion.nota for evaluacion in registro]
    promedio = sum(lista_notas) / len(lista_notas)

    if promedio >= 70.0:
        estado = "APROBADO"
    elif 60.0 <= promedio < 70.0:
        estado = "EN RIESGO"
    else:
        estado = "REPROBADO"

    print(f"Promedio acumulado actual: {promedio:.2f} / 100 pts")
    print(f"Estado de la asignatura: [{estado}]")
```

## Función 4: Generar Métricas de Rendimiento

Obtiene las calificaciones máxima y mínima del registro utilizando las funciones nativas max() y min(). Además, genera una proyección que le indica al estudiante cuántos puntos exactos le faltan en su promedio general para alcanzar la nota mínima aprobatoria (70).


```python
def generar_metricas(registro):
    print("\n--- MÉTRICAS Y RENDIMIENTO DETALLADO ---")
    if not registro:
        print("No hay notas suficientes para generar métricas.")
        return

    lista_notas = [ev.nota for ev in registro]
    lista_nombres = [ev.nombre for ev in registro]

    nota_maxima = max(lista_notas)
    nota_minima = min(lista_notas)

    idx_max = lista_notas.index(nota_maxima)
    idx_min = lista_notas.index(nota_minima)
    promedio_actual = sum(lista_notas) / len(lista_notas)

    print(f"• Nota más alta: {nota_maxima:.2f} pts ({lista_nombres[idx_max]})")
    print(f"• Nota más baja: {nota_minima:.2f} pts ({lista_nombres[idx_min]})")

    if promedio_actual >= 70.0:
        print("• Proyección: ¡Ya tienes el promedio suficiente para aprobar la asignatura!")
    else:
        puntos_faltantes = 70.0 - promedio_actual
        print(f"• Proyección: Te faltan {puntos_faltantes:.2f} pts en tu promedio para aprobar (70 pts).")
```

#### 4. BUCLE PRINCIPAL (MENÚ)

```python
def ejecutar_sistema():
    while True:
        print("\n==========================================")
        print(" SISTEMA DE NOTAS ACADÉMICAS (VERSIÓN POO) ")
        print("==========================================")
        print("1. Ver notas registradas (Datos iniciales)")
        print("2. Registrar una nueva nota")
        print("3. Calcular promedio y estado académico")
        print("4. Generar métricas de rendimiento")
        print("5. Salir")

        try:
            opcion = int(input("\nSeleccione una opción (1-5): "))

            if opcion == 1:
                mostrar_notas_precargadas(registro_academico)
            elif opcion == 2:
                registrar_nota(registro_academico)
            elif opcion == 3:
                calcular_promedio_y_estado(registro_academico)
            elif opcion == 4:
                generar_metricas(registro_academico)
            elif opcion == 5:
                print("\nSistema finalizado. ¡Éxito en la materia!")
                break
            else:
                print(" Opción no válida. Ingrese un número del 1 al 5.")

        except ValueError:
            print(" Error: Ingrese únicamente el número de la opción elegida.")
```

#### 5. PUNTO DE ENTRADA DEL SCRIPT

```python
if __name__ == "__main__":
    ejecutar_sistema()
```
