# Sistema de Control y Cálculo de Promedio Académico
**Asignatura:** Programación con Estructuras de Datos  
**Descripción:** Carga inicial de datos precargados mediante listas paralelas.

```python
evaluaciones = ["Tarea 1", "Examen Parcial 1", "Sistemático 1"]
notas = [85.0, 68.0, 92.0]

print("Datos precargados en el sistema exitosamente.")
```

### Fncion 1: Visualizar las notas registradas
Muestra en pantalla el desglose actual de las evaluaciones registradas en la asignatura.

```python
def mostrar_notas_precargadas(lista_evaluaciones, lista_notas):
    print("Evaluaciones y Notas Precargadas:")
    if not lista_evaluaciones or not lista_notas:
        print("No hay datos precargados.")
        return
    
    for i in range(len(lista_evaluaciones)):
        print(f"{lista_evaluaciones[i]}: {lista_notas[i]}")
```

### Funcion 2: Registgtrar nueva evaluacion
Permite ingresar un nuevo trabajo/examen.
Implementa maje de excepciones (try except) y validacion de rango (0 a 100)

```python
def registrar_nota(lista_evaluaciones, lista_notas):
    print("\n--- REGISTRAR NUEVA EVALUACIÓN ---")
    nombre_evaluacion = input(
        "Ingrese el nombre de la evaluación (ej. Tarea 2): "
    ).strip()

    if not nombre_evaluacion:
        print("Error: El nombre de la evaluación no puede estar vacío.")
        return

    while True:
        try:
            nota_input = float(input("Ingrese la nota obtenida (0 - 100): "))

            if 0 <= nota_input <= 100:
                lista_evaluaciones.append(nombre_evaluacion)
                lista_notas.append(nota_input)
                print(
                    f" Éxito: '{nombre_evaluacion}' guardada correctamente con"
                    f" {nota_input:.2f} pts."
                )
                break
            else:
                print(
                    " Error: La nota debe estar entre 0 y 100. Intente de"
                    " nuevo."
                )

        except ValueError:
            print(
                " Error de entrada: Por favor, ingrese un número válido (ej."
                " 85 o 78.5)."
            )
            
```

### Función 3: Calcular Promedio y Estado Académico
Procesa las calificaciones acumuladas, determina el promedio aritmético y evalúa el estado del estudiante (Aprobado, En Riesgo, Reprobado).

```python
def calcular_promedio_y_estado(lista_notas):
    print("\n--- BALANCE ACADÉMICO Y ESTADO ---")
    if not lista_notas:
        print("No hay notas registradas para calcular el promedio.")
        return
    
    promedio = sum(lista_notas) / len(lista_notas)
    
    if promedio >= 70.0:
        estado = "Aprobado"
    elif 60.0<= promedio < 70.0:
        estado = "En riesgo"
    else:
        estado = "Reprobado"
    print(f"Pomedio acumulado: {promedio:.2f}/100 puntos")
    print(f"Estado académico: {estado}")
```

### Función 4: Métricas de Rendimiento
Identifica la calificación máxima, mínima y calcula los puntos faltantes para alcanzar el promedio mínimo de aprobación (70 pts).

```python
def generar_metricas(lista_evaluaciones, lista_notas):
    print("\n METRICAS Y RENDIMIENTO DETALLADO")
    if not lista_notas:
        print("No hay notas suficientes para generar métricas.")
        return

    nota_maxima = max(lista_notas)
    nota_minima = min(lista_notas)

    idx_max = lista_notas.index(nota_maxima)
    idx_min = lista_notas.index(nota_minima)

    promedio_actual = sum(lista_notas) / len(lista_notas)

    print(
        f"• Nota más alta: {nota_maxima:.2f} pts"
        f" ({lista_evaluaciones[idx_max]})"
    )
    print(
        f"• Nota más baja: {nota_minima:.2f} pts"
        f" ({lista_evaluaciones[idx_min]})"
    )

    if promedio_actual >= 70.0:
        print(
            "• felicidades: Ya tienes el promedio suficiente para aprobar la"
            " asignatura"
        )
    else:
        puntos_faltantes = 70.0 - promedio_actual
        print(
            f" resultado: Te faltan {puntos_faltantes:.2f} pts en tu promedio"
            " acumulado para alcanzar la nota mínima de aprobación (70 pts)."
        )
        
```

### Menu principal de control 
Bucle de ejecución continua interactiva que integra las funciones anteriores

```python
def ejecutar_sistema():
    while True:
        print("\n==========================================")
        print("  SISTEMA DE CONTROL DE NOTAS ACADÉMICAS  ")
        print("==========================================")
        print("1. Ver notas registradas (Datos iniciales)")
        print("2. Registrar una nueva nota")
        print("3. Calcular promedio y estado académico")
        print("4. Generar métricas de rendimiento")
        print("5. Salir")

        try:
            opcion = int(input("\nSeleccione una opción (1-5): "))

            if opcion == 1:
                mostrar_notas_precargadas(evaluaciones, notas)
            elif opcion == 2:
                registrar_nota(evaluaciones, notas)
            elif opcion == 3:
                calcular_promedio_y_estado(notas)
            elif opcion == 4:
                generar_metricas(evaluaciones, notas)
            elif opcion == 5:
                print("\nSistema finalizado.")
                break
            else:
                print(" Opción no válida. Ingrese un número del 1 al 5.")

        except ValueError:
            print(" Error: Ingrese únicamente el número de la opción elegida.")

# Ejecución del sistema
ejecutar_sistema()
```
