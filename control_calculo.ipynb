Muestra en pantalla el desglose actual de las evaluaciones registradas en la asignatura.

```python
def mostrar_notas_precargadas(lista_evaluaciones, lista_notas):
    print("Evaluaciones y Notas Precargadas:")
    if not lista_evaluaciones or not lista_notas:
        print("No hay datos precargados.")
        return
    
    for i in range(len(lista_evaluaciones)):
        print(f"{lista_evaluaciones[i]}: {lista_notas[i]}")
```
